```python
import pandas as pd
import os

# Load the merged dataset
df = pd.read_csv("C:/Users/Durga Vishalini/diabetes-risk-prediction/data/modeling_dataset.csv")

# Step 1: One-Hot Encode Categorical Variables
categorical_cols = ["race", "gender", "age", "admission_type", "discharge_disposition", "admission_source"]
df = pd.get_dummies(df, columns=categorical_cols, drop_first=True)

# Step 2: Feature Engineering – Calculate Length of Stay
df['admission_date'] = pd.to_datetime(df['admission_date'])
df['discharge_date'] = pd.to_datetime(df['discharge_date'])
df['length_of_stay'] = (df['discharge_date'] - df['admission_date']).dt.days

# Drop unnecessary columns
df.drop(columns=["admission_date", "discharge_date", "encounter_id", "visit_id"], inplace=True)

# Step 3: Reorder columns so target 'readmitted' is at the end
target = 'readmitted'
cols = [col for col in df.columns if col != target] + [target]
df = df[cols]

# Step 4: Save to CSV for modeling
os.makedirs("data", exist_ok=True)
df.to_csv("C:/Users/Durga Vishalini/diabetes-risk-prediction/data/final_modeling_dataset.csv", index=False)

print("Feature engineering completed and saved to data/final_modeling_dataset.csv")
```
