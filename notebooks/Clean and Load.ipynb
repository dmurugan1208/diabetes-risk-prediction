# Validating keys

```python
import pandas as pd

# -------------------- LOAD CSVs --------------------
patients = pd.read_csv("C:/Users/Durga Vishalini/diabetes-risk-prediction/data/patients.csv")
visits = pd.read_csv("C:/Users/Durga Vishalini/diabetes-risk-prediction/data/hospital_visits.csv")
lab_results = pd.read_csv("C:/Users/Durga Vishalini/diabetes-risk-prediction/data/lab_results.csv")
medications = pd.read_csv("C:/Users/Durga Vishalini/diabetes-risk-prediction/data/medications.csv")

# -------------------- A. VALIDATE patient_nbr in visits --------------------
invalid_patient_nbrs = visits[~visits['patient_nbr'].isin(patients['patient_nbr'])]

if not invalid_patient_nbrs.empty:
    print("Visits with invalid patient_nbr (not found in patients):")
    print(invalid_patient_nbrs)
else:
    print("All patient_nbr values in hospital_visits exist in patients.csv")

# -------------------- B. VALIDATE encounter_id in lab_results and medications --------------------
invalid_encounters_lab = lab_results[~lab_results['encounter_id'].isin(visits['visit_id'])]
invalid_encounters_meds = medications[~medications['encounter_id'].isin(visits['visit_id'])]

if not invalid_encounters_lab.empty:
    print("Lab results with invalid encounter_id (not found in hospital_visits):")
    print(invalid_encounters_lab)
else:
    print("All encounter_id values in lab_results exist in hospital_visits.csv")

if not invalid_encounters_meds.empty:
    print("Medications with invalid encounter_id (not found in hospital_visits):")
    print(invalid_encounters_meds)
else:
    print("All encounter_id values in medications exist in hospital_visits.csv")
```

##  Why We Perform Key Validation
Key validation is an important step to make sure our data is correct and all tables are properly linked. Here's why we do it:

1. Ensure Correct Links Between Tables
We check that:

  *Every patient_nbr in the hospital_visits table exists in the patients table.
  
  *Every encounter_id in the lab_results and medications tables matches a visit_id in hospital_visits.
  
  This makes sure that all hospital visits belong to real patients, and all lab and medication records are connected to real       visits.

2. Avoid Missing or Broken Data
If these links are missing, some data may be left out when we join tables later. That could lead to wrong analysis or models that don’t work well.

3. Build Trustworthy Results
By checking the keys, we can confidently use the data for analysis, dashboards, or machine learning. It helps us avoid problems later and makes our system more reliable.
