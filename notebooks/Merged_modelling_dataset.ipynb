```python
import pandas as pd


base_path = "C:/Users/Durga Vishalini/diabetes-risk-prediction/data/"

patients = pd.read_csv(base_path + "patients.csv")
hospital_visits = pd.read_csv(base_path + "hospital_visits.csv")
lab_results = pd.read_csv(base_path + "lab_results.csv")
medications = pd.read_csv(base_path + "medications.csv")

# --- Step 1: Aggregate features ---
# Count lab results per encounter
lab_agg = lab_results.groupby("encounter_id").size().reset_index(name="num_lab_tests")

# Count medications per encounter
meds_agg = medications.groupby("encounter_id").size().reset_index(name="num_medications")

# --- Step 2: Merge all tables ---
# Merge patients + hospital_visits on patient_nbr
merged = pd.merge(patients, hospital_visits, on="patient_nbr", how="inner")

# Merge with lab_agg on encounter_id
merged = pd.merge(merged, lab_agg, on="encounter_id", how="left")

# Merge with meds_agg on encounter_id
merged = pd.merge(merged, meds_agg, on="encounter_id", how="left")

# Fill missing counts with 0
merged["num_lab_tests"] = merged["num_lab_tests"].fillna(0).astype(int)
merged["num_medications"] = merged["num_medications"].fillna(0).astype(int)

merged.to_csv("C:/Users/Durga Vishalini/diabetes-risk-prediction/data/modeling_dataset.csv", index=False)
print("Saved as 'data/modeling_dataset.csv'")
```
