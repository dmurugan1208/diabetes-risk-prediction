```python
import pandas as pd

# Load the merged dataset
df = pd.read_csv("C:/Users/Durga Vishalini/diabetes-risk-prediction/data/modeling_dataset.csv")

# Step 1: One-Hot Encoding
categorical_cols = ["race", "gender", "age"]
df = pd.get_dummies(df, columns=categorical_cols, drop_first=True)

# Step 2: Feature Engineering – Calculate Length of Stay
df['admission_date'] = pd.to_datetime(df['admission_date'])
df['discharge_date'] = pd.to_datetime(df['discharge_date'])
df['length_of_stay'] = (df['discharge_date'] - df['admission_date']).dt.days

# Drop original date columns
df.drop(["admission_date", "discharge_date"], axis=1, inplace=True)

# Save the updated dataset
df.to_csv("C:/Users/Durga Vishalini/diabetes-risk-prediction/data/processed_modeling_dataset.csv", index=False)

print("Feature engineering completed and saved as 'processed_modeling_dataset.csv'")
```
